**5-HISTOGRAM**

```python
#data
#used for distrbution analysis
import random #Iska use random numbers generate karne ke liye hota hai.
data=[random.randint(1,20) for _ in range (600)]#1 se 20 ke beech random integer deta hai.
#range(600) → Loop ko 600 baar chalata hai.
plt.hist(data,bins=20)#bins is used for how many part u want to divide data
plt.plot()
```

**6-pie chart**

```python
#use to show part to whole relationship/data
categories=['A','B','C','D','E']
MARKS=[20,30,40,50,48]
plt.pie(MARKS, labels=categories, autopct="%1.1f%%")
#("%1.1f%%")=Har slice ka percentage ek decimal place tak dikhata hai.
plt.title("marks chart")
plt.show()
```

```python
#use to show part to whole relationship/data
categories=['A','B','C','D','E']
MARKS=[20,30,40,50,48]
plt.pie(MARKS, labels=categories, autopct="%1.1f%%",startangle=90)# chart will start from 90 dgree
#("%1.1f%%")=Har slice ka percentage ek decimal place tak dikhata hai.
plt.title("marks pie chart")
plt.show()
```

**7-Scatter plot**

```python
#used to find relationship btw variable
x=[1,2,3,4]
y=[40,35,50,60]
plt.scatter(x,y,color='r')
plt.title("scatter plot chart")
plt.show()
```

```python
#data 1
section=['A','B','C','D','E']
marks=[20,30,40,50,48]
#data 2
x=[1,2,3,4]
y=[40,35,50,60]
plt.figure(figsize=(8,4))
#first plot-bar chart
plt.subplot(1,2,1)#row , col. ,position
plt.bar(section,marks)
plt.title("marks")
plt.xlabel("section")
plt.ylabel("marks")
#sec plot scatter chart
plt.subplot(1,2,2)#row , col. ,position
plt.scatter(x,y)
plt.title("marks")
plt.xlabel("section")
plt.ylabel("marks")
plt.show()
```
