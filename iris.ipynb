```python
import matplotlib.pyplot as plt
```

```python
plt.bar(df["species"], df["sepal_length"])
plt.xlabel("Species")
plt.ylabel("Sepal Length")
plt.title("Sepal Length by Species")
plt.show()
```

```python
plt.bar(df["species"], df["petal_length"], label="Petal Length")
plt.bar(df["species"], df["petal_width"], label="Petal Width")

plt.xlabel("Species")
plt.ylabel("Value")
plt.title("Petal Length and Petal Width")
plt.legend()
plt.show()
```
